```python
x = np.linspace(0,3,100)
```

```python
def y(arr):
    return -(arr**2) + 3*arr - 1
```

```python
y = y(x)
```

```python
plt.plot(x,y)
plt.show()
```

```python
for idx,val in enumerate(y) :
    if val == y.max():
        print(f'Y max: {val}')
        print(f'Index y :{idx}')
        print(f'Index x :{x[idx]}')
```
