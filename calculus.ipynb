```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
sns.set(style='darkgrid')

def draw_secant(x_values):
    x = np.linspace(-20,30,100)
    y = -1*(x**2) + x*3 - 1
    plt.plot(x,y)
    # Add your code here.
    x1 = x_values[0]
    y1 = -1*(x1**2) + x1*3 - 1
    x2 = x_values[1]
    y2 = -1*(x2**2) + x2*3 - 1
    m = (y1-y2)/(x1-x2)
    b = y1 - m*x1
    y_secant = m*x + b
    plt.plot(x,y_secant,color='green')
    plt.show()

draw_secant([3,5])
```
